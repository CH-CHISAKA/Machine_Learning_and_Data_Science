# Introduction to Computer Vision

## What is Computer Vision?
Computer Vision (CV) is a field of Artificial Intelligence that enables computer to interpret, analyse and understand visual information.

This information can come from:
* Digital photographs
* Videos 
* Medical scans
* Satellite images.
* CCTV cameras
* Real-time camera feeds

_In simple terms_:

Computer Vision teaches computers to see and understand images in a way that allows them to make decisions.

- For example, when you take a photograph of an elephant, a computer vision model can analyse the pixels and predict that the image contains an elephant.

## The three core Computer Vision tasks

You must understand the difference between 
* Classification, 
* Detection 
* Segmentation.


##### **Core Computer Vision Tasks**

* **Object Detection** - _(standard axis-aligned/horizontal bounding box, often called HBB or AABB)_: Locates and classifies multiple objects in images or video streams using bounding boxes.

* **Instance & Semantic Segmentation**: Generates pixel-level masks for individual objects (instance) or labels every pixel by class (semantic) for detailed scene understanding and background editing.

* **Oriented Object Detection (OBB)**: Uses rotated bounding boxes to tightly fit angled or elongated objects, such as vehicles, ships, or aerial targets.

* **Pose Estimation**: Detects key body joints or landmarks to track human or animal movement and posture.


* **Image Classification & Depth Estimation**: Categorizes entire images into classes or estimates per-pixel depth from a single RGB image.

* **Object Tracking**: Follows detected objects across consecutive video frames to monitor movement and counts.

### Image Classification

**Definition**: Image classification involves assigning a category or label to an entire image.

The model receives an image and predicts what it contains.

Example:

Input: Photograph of an elephant.

Output:
- Elephant: 90%
- Rhinoceros: 7%
- Hippopotamus: 3%

These percentages represent the model's predicted class probabilities.

How it works:

1. The image is provided to the model.
2. The model extracts visual features.
3. It analyses these features.
4. It produces probabilities for possible categories.
5. The class with the highest probability becomes the prediction.

<img src="img/example.png" alt="Image Classification" width="500">

_Practical example_: 

A manufacturing company classifies products as defective or non-defective.

### Object Detection 
**Definition**: Object detection identifies individual objects in an image and determines their locations.

Unlike classification, detection tells us both:

- What objects are present?
- Where are the objects located?

The output includes:

- Class labels.
- Bounding boxes.
- Confidence scores.

_A bounding box is a rectangular box drawn around a detected object._


For example, an image containing three cars could produce:

- Car 1: 96% confidence.
- Car 2: 92% confidence.
- Car 3: 89% confidence.

Each car receives its own bounding box.


<img src="img/example.png" alt="Object Detection" width="500">

A popular detection algorithm is **YOLO (You Only Look Once)**.

_Practical example_: A self-driving car detects pedestrians, traffic signs and other vehicles.

### Image Segmentation
**Definition**: Image segmentation assigns a category to individual pixels in an image.

Rather than simply identifying an object or drawing a rectangle around it, segmentation identifies the actual pixels belonging to the object.

- For example, in a medical scan, segmentation can identify the exact region occupied by a tumour.

**Two important types:**

1. Semantic Segmentation 

2. Instance Segmentation

#### Semantic Segmentation

Every pixel is assigned a class.

For example:
- Road pixels.
- Car pixels.
- Pedestrian pixels.
- Building pixels.

Objects belonging to the same category receive the same class label.
- a computer vision task that assigns a category label (such as road, sky, tree, or person) to every individual pixel in an 

image. It produces a dense pixel-level class map rather than a single image label or bounding

#### Instance Segmentation

Identifies individual objects separately, even when they belong to the same category.

**For example,** if an image contains five cars, each car receives its own separate mask.

**Comparison**

| Feature | Classification | Detection | Segmentation |
| --- | --- | --- | --- |
| Main purpose | Identify image category | Identify and locate objects | Identify exact object pixels |
| Output | Class label | Bounding boxes | Pixel masks |
| Location information | No | Yes, approximate | Yes, precise |
| Complexity | Relatively simple | More complex | More detailed |
| Example | Is this a cat? | Where are the cats? | Which pixels form each cat? |

Exam tip: Remember this distinction:

- Classification = What is in the image?

- Detection = What is in the image, and where is it?

- Segmentation = Which exact pixels belong to each object?

## Understanding Digital Images

Before implementing computer vision models, you need to understand how computers represent images.

### Images are arrays of numbers**

A digital image is made up of tiny elements called pixels _(picture elements)_.

<img src="img/pixels.png" alt="Picture Elements" width="500">
<p><strong>Pixels (Picture ELements)</strong></p>


- Each pixel stores numerical information representing its colour or intensity.

For example, a grayscale image uses one value per pixel:

- 0 = Black.

- 255 = White.

- Values between 0 and 255 = Different shades of grey.

<img src="img/understanding_digital_images.png" alt="Understanding Digital Images" width="500">
<p><strong>Understanding Digital Images</strong></p>

A colour RGB image uses three channels:

- R = Red.

- G = Green.

- B = Blue.

Each channel has a value between 0 and 255.

<img src="img/understanding_digital_images_rgb.png" alt="Understanding Digital Images RGB" width="500">
<p><strong>Understanding Digital Images RGB</strong></p>

### mage dimensions
An image is commonly represented using the following dimensions:
$$
(H,W,C)
$$

Where:

- H = Height.

- W = Width.

- C = Number of colour channels.

| Image Type | Shape |
| --- | --- |
| Grayscale 28 × 28 | `(28, 28)` |
| Grayscale with channel | `(28, 28, 1)` |
| RGB 224 × 224 | `(224, 224, 3)` |


A batch of 32 RGB images of size 224 × 224 would have the shape:

$$
(32,224,224,3)
$$

The first dimension represents the number of images.
