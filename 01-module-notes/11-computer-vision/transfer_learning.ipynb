# Transfer Learning

## Understanding Transfer Learning

Transfer Learning is one of the most important concepts in practical machine learning.

**Definition**: 

Transfer learning involves taking knowledge learned by a model on one task and reusing it to solve another related task.

For example:

- Original task: Recognising thousands of ImageNet categories.

- New task: Recognising clothing categories.

Instead of training an entirely new model, we reuse the features learned by the existing model.

### Why does transfer learning work?

Many visual features are reusable.

A model trained on animals, vehicles and everyday objects may already understand:

- Edges.

- Shapes.

- Textures.

- Patterns.

- Object structures.

These features can also help recognise clothing, even if the model has never been specifically trained to classify Fashion-MNIST clothing categories.

## Two important approaches

1. Feature Extraction 
2. Fine Tuning

### A. Feature extraction

Use the pretrained model as a fixed feature extractor.

- Keep the pretrained layers frozen.

- Replace the original classification head.

- Train only the new classification layers.

### B. Fine-tuning

Allow some of the pretrained layers to continue learning.

- Unfreeze selected layers.

- Use a small learning rate.

- Train the model on the new dataset.

**Feature extraction vs fine-tuning**

| Feature | Feature extraction | Fine-tuning |
| --- | --- | --- |
| Pretrained layers | Frozen | Some are trainable |
| New classifier | Trained | Trained |
| Training time | Usually shorter | Usually longer |
| Computational cost | Lower | Higher |
| Risk of overfitting | Generally lower | Can be higher |
| Adaptation | Limited | Greater |

**Easy way to remember:**

- Feature extraction = Reuse existing knowledge without changing it.

- Fine-tuning = Reuse existing knowledge and adjust some of it to fit the new task.
