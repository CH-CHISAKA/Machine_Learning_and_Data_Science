# Vision Transformers (ViT)

## Introduction to Vision Transformers

An alternative to traditional CNN architectures.

### Why were Vision Transformers introduced?
CNNs are very good at identifying local visual patterns.

However, CNNs generally combine information progressively:
1. Detect simple patterns.

2. Combine them into more complex patterns.

3. Build increasingly larger representations.

### Vision Transformers approach this differently.

- They divide an image into small patches and process these patches as tokens.

- They use a mechanism called self-attention to understand relationships between different patches.

## Understanding image patches
Imagine an image measuring:
$$
28\times28
$$

Instead of processing the entire image as one structure, a Vision Transformer divides it into smaller patches.

For example, using patch size 4:
$$
\frac{28}{4}=7
$$

There are seven patches horizontally and seven vertically.

Total number of patches:
$$
7\times7=49
$$

Each patch is converted into a numerical representation called an embedding.

- The embeddings are then processed by the transformer.

## What is a token?
In Natural Language Processing (NLP), a token represents a word or part of a word.

In Vision Transformers, a token represents an image patch.

Therefore:
- NLP: Sentence → Words → Tokens.

- ViT: Image → Patches → Tokens.

## Positional encoding
When an image is divided into patches, the model needs to understand where each patch originally came from.

For example, the position of an eye in an image is important.

Positional encoding adds information about the location of each patch.

Without positional information, the model would have difficulty preserving the spatial arrangement of the image.

#### Self-attention
Self-attention allows each patch to consider information from other patches.

For example, when identifying a shoe:
- One patch may contain the sole.

- Another may contain the upper section.

- Another may contain the heel.

Self-attention allows the model to understand how these regions relate to one another.

This is especially useful for understanding global image structure.

**CNN vs Vision Transformer**

| Feature | CNN | Vision Transformer |
| --- | --- | --- |
| Input processing | Convolutional operations | Image patches |
| Main mechanism | Convolution | Self-attention |
| Feature learning | Local patterns progressively combined | Relationships between patches |
| Spatial information | Built into convolutional structure | Positional embeddings |
| Global relationships | Built progressively | Directly modelled through attention |
| Architecture | Convolutional layers | Transformer blocks |

**Remember**: CNNs learn local features progressively, whereas ViTs use self-attention to model relationships between image patches.

## Understanding Swin Transformers
Your lecture also introduces Swin Transformers.

A standard Vision Transformer uses global self-attention.

This can become computationally expensive as the number of image patches increases.

Swin Transformers address this using window-based attention.
How does Swin work?
1. Divide the image into patches.

2. Group patches into local windows.

3. Apply attention within each window.

4. Shift the windows in subsequent layers.

5. Allow information to flow between neighbouring windows.
