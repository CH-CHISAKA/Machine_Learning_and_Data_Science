# Introduction to Computer Vision

## What is Computer Vision?
Computer Vision (CV) is a field of Artificial Intelligence that enables computer to interpret, analyse and understand visual information.

This information can come from:
* Digital photographs
* Videos 
* Medical scans
* Satellite images.
* CCTV cameras
* Real-time camera feeds

_In simple terms_:

Computer Vision teaches computers to see and understand images in a way that allows them to make decisions.

- For example, when you take a photograph of an elephant, a computer vision model can analyse the pixels and predict that the image contains an elephant.

## The three core Computer Vision tasks

You must understand the difference between 
* Classification, 
* Detection 
* Segmentation.


##### **Core Computer Vision Tasks**

* **Object Detection** - _(standard axis-aligned/horizontal bounding box, often called HBB or AABB)_: Locates and classifies multiple objects in images or video streams using bounding boxes.

* **Instance & Semantic Segmentation**: Generates pixel-level masks for individual objects (instance) or labels every pixel by class (semantic) for detailed scene understanding and background editing.

* **Oriented Object Detection (OBB)**: Uses rotated bounding boxes to tightly fit angled or elongated objects, such as vehicles, ships, or aerial targets.

* **Pose Estimation**: Detects key body joints or landmarks to track human or animal movement and posture.


* **Image Classification & Depth Estimation**: Categorizes entire images into classes or estimates per-pixel depth from a single RGB image.

* **Object Tracking**: Follows detected objects across consecutive video frames to monitor movement and counts.
