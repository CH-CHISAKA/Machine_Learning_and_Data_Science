# Question

##### **How to estimate how well a model performs on new/unseen data.**

Model use case = **_Linear Regression_**

# Pseudocode

1. Load the dataset 

2. Shuffle the data randomly 

3. Split the data into `k` equal-sized groups called **fold**

4. For each fold i from 1 to k:

* Use fold i as the test set.

* Use all the other folds as the training set.

* Train the linear regression model using the training set.

* Test the model on fold i and record its performance.

5. Repeat this process until every fold has been used as the test set once.

6. Calculate the average of the k performance scores.

7. Use this average as the estimated performance of the model.


#### Formula 
$$
Split → Train → Test → Record → Repeat → Average
$$

# Example 

With _**5-fold cross-validation**_, the data is split into 5 parts 

* Round 1: Fold 1 = test, Folds 2–5 = training

* Round 2: Fold 2 = test, Folds 1, 3-5 = training

* Round 3: Fold 3 = test, Folds 1, 2, 4 and 5 = training

* Round 4: Fold 4 = test, Folds 1, 2, 3 and 5 = training

* Round 5: Fold 5 = test, Folds 1-4 = training

**Average the 5 test scores** to get the cross-validation estimate.
