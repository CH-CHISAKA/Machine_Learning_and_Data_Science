# 01: Importing Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

# 02: Loading and Display the Palmer Penguins Dataset

```python
# Loading the palmer penguins dataset
penguins = sns.load_dataset("penguins")

# Displaying the head(5) - first five row, and tail(5) last five rows of the dataset
print("Head of the dataset:")
display(penguins.head(5))

print("\nTail of the dataset:")
display(penguins.tail(5))
```

```python
# The size of the dataset
print(f"\nRows and Columns: {penguins.shape}")

# Column information and data types
print("\nColumn information and data types:")
display(penguins.info())
```

```python
# Checking for missing values in the dataset
print("\nMissing values in the dataset:")
display(penguins.isnull().sum())
```

```python
# Statistical summary
print("\nStatistical summary of the dataset:")
display(penguins.describe(include='all'))
```

# 03: Explore variables

## Bill Length

```python
plt.figure(figsize=(12,8))
sns.histplot(penguins["bill_length_mm"], bins=20, kde=True, color="steelblue")
plt.title("Distribution of Bill Length")
plt.tight_layout()
plt.show()
```

**_Summary_**

* A bimodal distribution; from around _39–40 mm_ and _46–51 mm_. An observable dip is noticeable around _43–44 mm_. 

* Most penguins have bill lengths between _35–50 mm_.

* Slightly skewed to the right.

## Bill Depth

```python
plt.figure(figsize=(12,8))
sns.histplot(penguins["bill_depth_mm"], bins=20, kde=True, color="royalblue")
plt.title("Distribution of Bill Depth")
plt.tight_layout()
plt.show()
```

**_Summary_**

- The distribution is moderately spread out, with most bill depths falling between about _15 and 20 mm_

- Fairly concentrated around _17–19 mm_, where most observations occur.

- There is some evidence of multiple groups, as shown by the two noticeable peaks.

## Flipper Length

```python
plt.figure(figsize=(12,8))
sns.histplot(penguins["flipper_length_mm"], bins=20, kde=True, color="teal")
plt.title("Distribution of Flipper Length")
plt.tight_layout()
plt.show()
```

**_Summary_**

- The distribution is bimodal, with two noticeable groups rather than one smooth peak.

- Most flipper lengths are concentrated around _187–215 mm_.

- First group: A large concentration occurs around _185–200 mm_, with a peak near _190–195 mm_.

- Second group: Another concentration appears around _207–220 mm_, peaking near _210–215 mm_.

- _Gentoo penguins_ contain relatively larger bill dimensions, which can contribute to the generally longer flippers, reflecting the difference between the penguin species.

## Body Mass

```python
plt.figure(figsize=(12,8))
sns.histplot(penguins["body_mass_g"], bins=20, kde=True, color="darkblue")
plt.title("Distribution of Body Mass")
plt.tight_layout()
plt.show()
```

**_Summary_**

- A _Unimdoal Distribution_, with a _positive, right skewness_, with a clear main peak and longer tail towards, higher body masses.

- Most body masses are concentrated around _3,500–4,500 g_.

- he largest concentration occurs around _3,600–3,800 g_, with the highest frequency near _3,700 g_.

- There are a few smaller bumps around 4,500–5,500 g, but they are not distinct enough to classify the distribution as _multimodal_.



# 04: Compare Relationships

## Correlation Heatmap

```python
numeric_cols = penguins.select_dtypes(include="number")

plt.figure(figsize=(8,6))
sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

**_Summary_**

* Strong positive correlation between:

- Flipper length and body mass

- Bill length and flipper length

- Weak or negative correlations between some bill measurements; i.e. bill depth.

## Scatter Plot: Bill Length vs Bill Depth

```python
plt.figure(figsize=(12,8))
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    s=80
)
plt.title("Bill Length vs Bill Depth")
plt.tight_layout()
plt.show()
```

**_Summary_**

- Species form distinct clusters.

- Adelie tends to have shorter but deeper bills.

- Chinstrap tends to have longer bills.

- Gentoo tends to have shallower bills.

**_Bill length_** and **_Bill depth_** can help distinguish penguin species.



## Scatter Plot: Flipper Length vs Body Mass

```python
plt.figure(figsize=(12,8))
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    s=80
)
plt.title("Flipper Length vs Body Mass")
plt.tight_layout()
plt.show()
```

**_Summary_**

- Gentoo penguins are generally the largest.

- Larger flippers can be associated with greater body mass, therefore a _Strong positive Relationship_.

## Boxplot: Body Mass by Species

```python
plt.figure(figsize=(12,8))
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
)
plt.title("Body Mass by Species")
plt.tight_layout()
plt.show()
```

**_Summary_**

- Gentoo penguins have the highest median body mass.

- Adelie penguins are the lightest.

- Some overlap exists, but species differences are clear.

## Boxplot: Body Mass by Sex

```python
plt.figure(figsize=(12,8))
sns.boxplot(
    data=penguins,
    x="sex",
    y="body_mass_g"
)
plt.title("Body Mass by Sex")
plt.tight_layout()
plt.show()
```

**_Summary_**

- The _Male penguins_ generally have a higher body mass than _Female penguins_; the distribution of males is shifted upwards.
