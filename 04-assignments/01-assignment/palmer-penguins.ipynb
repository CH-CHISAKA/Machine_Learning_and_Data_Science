# 01: Importing Libraries

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
````

# 02: Loading and Display the Palmer Penguins Dataset

````python
# Loading the palmer penguins dataset
penguins = sns.load_dataset("penguins")

# Displaying the head(5) - first five row, and tail(5) last five rows of the dataset
print("Head of the dataset:")
display(penguins.head(5))

print("\nTail of the dataset:")
display(penguins.tail(5))
````

````python
# The size of the dataset
print(f"\nRows and Columns: {penguins.shape}")

# Column information and data types
print("\nColumn information and data types:")
penguins.info()
````

````python
# Checking for missing values in the dataset
print("\nMissing values in the dataset:")
display(penguins.isnull().sum())
````

````python
# Species count
species_sex_counts = pd.crosstab(
    penguins["species"],
    penguins["sex"]
)

print(species_sex_counts)
````

````python
# Statistical summary
print("\nStatistical summary of the dataset:")
display(penguins.describe(include='all'))
````

# 
**Summary**

The dataset contains 3-three unique penguin species:
* Adelie - _146_, both male and female 50/50 split (_73_ per sex)
* Chinstrap - _68_, both male and female 50/50 split (_34_ per sex)
* Gentoo - _119_, with males taking _51.3_% and females _48.7_%

The dataset contains the following characteristics, 344 observations and 7 variables; (3 Categorical - _species_, _island_, and _sex_; and 4 Numerical - _bill measurements_, _flipper length_, _body mass_, and _bill depth_):


Some variables/characteristics such as _sex_, _flipper length_, _body mass_, contain missing value (NaN), with _sex_ having the most missing values **11**.

# 03: Explore variables

## Bill Length

````python
plt.figure(figsize=(12,8))
sns.histplot(penguins["bill_length_mm"], bins=20, kde=True, color="steelblue")
plt.title("Distribution of Bill Length")
plt.tight_layout()
plt.show()
````

**_Summary_**
* The Bill length ranges from _32.1_ to _59.6_ mm.
    - The mean is _43.92_ mm and a standard deviation of _5.46_ mm.

* A bimodal distribution; from around _39–40 mm_ and _46–51 mm_. An observable dip is noticeable around _43–44 mm_. 

* Most penguins have bill lengths between _35–50 mm_.

## Bill Depth

````python
plt.figure(figsize=(12,8))
sns.histplot(penguins["bill_depth_mm"], bins=20, kde=True, color="royalblue")
plt.title("Distribution of Bill Depth")
plt.tight_layout()
plt.show()
````

**_Summary_**

- Bill depth ranges from 13.1 to 21.5 mm;
    * The mean of _17.15_ mm and a standard deviation of _1.97_ mm.

- The distribution is moderately spread out, with most bill depths falling between about _15 and 20 mm_

- The distribution is bimodal with the lower mode roughly _14–16_ mm and the upper mode around _17–19_ mm.

- Fairly concentrated around _17–19 mm_, where most observations occur.

- There is some evidence of multiple groups, as shown by the two noticeable peaks.

## Flipper Length

````python
plt.figure(figsize=(12,8))
sns.histplot(penguins["flipper_length_mm"], bins=20, kde=True, color="teal")
plt.title("Distribution of Flipper Length")
plt.tight_layout()
plt.show()
````

**_Summary_**
- The Flipper length ranges from _172 to 231_ mm, with a mean of _200.92_ mm and a standard deviation of _14.06_ mm.

- The distribution is bimodal, with two noticeable groups rather than one smooth peak.

- Most flipper lengths are concentrated around _187–215 mm_.

- First group: A large concentration occurs around _185–200 mm_, with a peak near _190–195 mm_.

- Second group: Another concentration appears around _207–220 mm_, peaking near _210–215 mm_.

- _Gentoo penguins_ contain relatively larger bill dimensions, which can contribute to the generally longer flippers, reflecting the difference between the penguin species.

## Body Mass

````python
plt.figure(figsize=(12,8))
sns.histplot(penguins["body_mass_g"], bins=20, kde=True, color="darkblue")
plt.title("Distribution of Body Mass")
plt.tight_layout()
plt.show()
````

**_Summary_**
- The Body mass ranges from _2,700 to 6,300_ g, with a mean of _4,201.75_ g and a standard deviation of _801.95_ g.

- A _Unimodal Distribution_, with a _positive, right skewness_, with a clear main peak and longer tail towards, higher body masses.

- Most body masses are concentrated around _3,500–4,500 g_, and a long right tail extension - a positive skewness.

- The largest concentration occurs around _3,600–3,800 g_, with the highest frequency near _3,700 g_.

- There are a few smaller bumps around _4,500–5,500_ g, but they are not distinct enough to classify the distribution as _multimodal_.



# 04: Compare Relationships

## Correlation Heatmap

````python
numeric_cols = penguins.select_dtypes(include="number")

plt.figure(figsize=(8,6))
sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
````

**_Summary_**

* Strong positive correlation between:

    - Flipper length and body mass (= _0.87_) — the strongest relationship in the dataset.

    - Bill length and flipper length (= _0.66_), moderate to strong
    
    - Bill length and body mass (= _0.60_) — moderate to strong.

* Moderate negative correlations:

    - Bill depth and flipper length (= _−0.58_), 

    - Bill depth and body mass (= _−0.47_).

    - Bill length and bill depth (= _−0.24_).

## Scatter Plot: Bill Length vs Bill Depth

````python
plt.figure(figsize=(12,8))
sns.scatterplot(
    data=penguins,
    x="bill_length_mm",
    y="bill_depth_mm",
    hue="species",
    s=80
)
plt.title("Bill Length vs Bill Depth")
plt.tight_layout()
plt.show()
````

**_Summary_**

- Species form distinct clusters, with slight overlap;

    - Adelie tends to have short bills _(mean 38.79 mm)_, that are deep _(18.35 mm)_ - **_lower left_**.

    - Chinstrap tends to have long bills _(48.83 mm)_, that are deep _(18.42 mm)_ — **_upper right_**.

    - Gentoo tends to have long bills _(47.50 mm)_ that are shallow _(14.98 mm)_ - **_lower right_**.

Interesting to find out that Chinstrap and Adelie are essentially identical in bill depth, and the bill length is what separates them.



## Scatter Plot: Flipper Length vs Body Mass

````python
plt.figure(figsize=(12,8))
sns.scatterplot(
    data=penguins,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    s=80
)
plt.title("Flipper Length vs Body Mass")
plt.tight_layout()
plt.show()
````

**_Summary_**
- There is a strong positive relationship between flipper length and body mass (= _0.87_): penguins with longer flippers are consistently heavier.

- Gentoo penguins are generally the largest, upper-right of the plot.

- Adelie and Chinstrap overlap heavily with each other, distinguishing Gentoo well but does not distinguish Adelie from Chinstrap.

- Larger flippers can be associated with greater body mass, therefore a _Strong positive Relationship_.

## Boxplot: Body Mass by Species

````python
plt.figure(figsize=(12,8))
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g",
)
plt.title("Body Mass by Species")
plt.tight_layout()
plt.show()
````

**_Summary_**

- Gentoo penguins have the highest median body mass.

- Adelie penguins are the lightest.

- Some overlap exists, but species differences are clear.

## Boxplot: Body Mass by Sex

````python
plt.figure(figsize=(12,8))
sns.boxplot(
    data=penguins,
    x="sex",
    y="body_mass_g"
)
plt.title("Body Mass by Sex")
plt.tight_layout()
plt.show()
````

**_Summary_**

- The _Male penguins_ generally have a higher body mass than _Female penguins_; the distribution of males is shifted upwards.

## Statistical Summary by Species

````python
species_summary = penguins.groupby("species")[
    ["bill_length_mm",
     "bill_depth_mm",
     "flipper_length_mm",
     "body_mass_g"]
].mean()

print(species_summary.round(2))
````

**_Summary_**
- Adelie
    * Smaller body size _3700.66 g_ and shortest flipper length _189.95 mm_, with deep bills (_18.35 mm_) and a mean body mass of _3,700.66 g_.

- Chinstrap
    * Has the longest bill length _48.83 mm_ as well as the bill depth of _18.42 mm_, and body mass of _3,733.09 g_.

- Gentoo
    * Has the highest body mass of _5076.02 g_, which has a strong positive linear relationship with the flipper length _217.19 mm_, with shallow bill _(14.98 mm)_.

## Statistical Summary by Sex

````python
sex_summary = penguins.groupby("sex")[
    ["bill_length_mm",
     "bill_depth_mm",
     "flipper_length_mm",
     "body_mass_g"]
].mean()

print(sex_summary.round(2))
````

**_Summary_**

- Males are generally larger across all measurements.

- The largest difference appears on the body mass male _4,545.68 g_ vs _3,862.27 g_ female

## Body Mass Distribution by Species

````python
plt.figure(figsize=(12,8))
sns.boxplot(
    data=penguins,
    x="species",
    y="body_mass_g"
)

plt.title("Body Mass Differs Across Penguin Species", fontsize=16)
plt.suptitle("Body Mass Distribution by Species", fontsize=22, y=1)
plt.xlabel("Species")
plt.ylabel("Body Mass (g)")
plt.tight_layout()
plt.show()
````

    * Adelie and Chinstrap penguins have lower average body masses, with almost similar bill depth and are separated mainly by bill length.

- The Flipper length and body mass are strongly positively related (_0.87_), both across the sample and within each species.

- The Exploratory analysis 
    * revealed clear differences among species in body size and bill characteristics.

- Both the Scatter and Box plot provide clear information that show species can be distinguished fairly well using body and bill measurements.

- The dataset provided context and good practice of how visualization and summary statistics can reveal relationships and differences that are difficult to see from raw data alone.
